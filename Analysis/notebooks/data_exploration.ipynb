```python
import pandas as pd

# Load dataset
df = pd.read_csv("c:/Dishant/Data Analysis/ecommerce-customer-analysis/data/raw_data/online_shoppers_intention.csv")

# Dataset overview
print("=" * 60)
print("DATASET OVERVIEW")
print("=" * 60)

print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])
print("Dataset Size:", round(df.memory_usage(deep=True).sum() / 1024, 2), "KB")

#Column information
print("\nCOLUMN NAMES:")
print(df.columns.tolist())

# Data types
print("\nDATA TYPES:")
print(df.dtypes)

# Dataset information
print("\nDATASET INFORMATION:")
df.info()

# Sample records
print("\nFIRST FIVE RECORDS:")
print(df.head())

# Target variable distribution
print("\nPURCHASE OUTCOME DISTRIBUTION:")
print(df["Revenue"].value_counts())

print("\nPURCHASE OUTCOME PERCENTAGE:")
print(
    (df["Revenue"].value_counts(normalize=True) * 100).round(2)
)

# Categorical values
print("\nVISITOR TYPES:")
print(df["VisitorType"].value_counts())

print("\nMONTHS:")
print(df["Month"].value_counts())

print("\nWEEKEND DISTRIBUTION:")
print(df["Weekend"].value_counts())

print("\nDATASET VERIFICATION COMPLETED")
```
